# Supabase

## Configurações

````python
import os
from dotenv import load_dotenv
from supabase import create_client, Client
load_dotenv()
SUPABASE_URL = os.getenv("SUPABASE_URL")
SUPABASE_KEY = os.getenv("SUPABASE_KEY")
````

````python
supabase: Client = create_client(
SUPABASE_URL,
SUPABASE_KEY
)
````

## Inserindo dados

````python
for verbo in verbos:
    response = supabase.table('verbs').insert({
        'root_verb': verbo['root_verb'],
        '3s_present': verbo['3s_present'],
        'past_simple': verbo['past_simple'],
        'present_participle': verbo['present_participle'],
        'past_participle': verbo['past_participle'],
        'verbo_base': verbo['verbo_base'],
        'presente': verbo['presente'],
        'passado': verbo['passado'],
        'gerundio': verbo['gerundio'],
        'participio_passado': verbo['participio_passado'],
        'futuro_subjuntivo': verbo['futuro_subjuntivo'],
        'pret_subjun_imperf': verbo['pret_subjun_imperf']
        }).execute()
    print(response.data)
````

## Consulta

````python
response = supabase.table('verbs').select("*").execute()
print(response.data)
dados_consulta  = response.data
````
